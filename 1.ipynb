```python
Attributs = np.array([[1,7,3],
                     [1/7,1,1/3],
                     [1/3,3,1]])
eigenvalues, eigenvectors = np.linalg.eig(Attributs)


real_mask = np.isreal(eigenvalues)
real_eigenvalues = eigenvalues[real_mask].real
real_eigenvectors = eigenvectors[:, real_mask].real


max_idx = np.argmax(real_eigenvalues)

biggest_eigenvalue = real_eigenvalues[max_idx]
corresponding_eigenvector = real_eigenvectors[:, max_idx]

print("Biggest Eigenvalue:", biggest_eigenvalue)
print("Corresponding Eigenvector:", corresponding_eigenvector) 



A1 = np.array([[1, 7],
               [0.143, 1]])
A2 = np.array([[1, 0.2],
               [5, 1]])
A3 = np.array([[1, 0.111],
               [9, 1]])

```

```python
import numpy as np

A = np.array([[30, 0.1, 1, 20],
              [100, 0.7, 1, 40],
              [50, 1, 2, 10],
              [300, 2, 3, 35]], dtype=float) # Force float to prevent integer truncation

for i in range(A.shape[1]):
    col_min = np.min(A[:, i])
    col_max = np.max(A[:, i])
    
    if i == 2:
        A[:, i] = (A[:, i] - col_min) / (col_max - col_min)
    else:
        A[:, i] = (col_max - A[:, i]) / (col_max - col_min)

columns_std = np.std(A, axis=0)
print("Standard deviation attribute:\n", columns_std)



print("\nNormalized Matrix:\n", A)

corr_matrix = np.corrcoef(A, rowvar=False)
print("\nCorrelation Matrix:\n", corr_matrix)

C = np.zeros(A.shape[1])
for j in range(A.shape[1]):
    conflict_sum = np.sum(1 - corr_matrix[j, :])
    C[j] = columns_std[j] * conflict_sum

print("\nConflict Coefficients:\n", C)

```
